```python
import pandas as pd

# Charger le fichier Excel
df = pd.read_excel("classif.xlsx")

# Compter la fréquence des bug_type
counts = df["bug type"].value_counts()

# Afficher les 3 classes les plus fréquentes
top_classes = counts.head(3).index.tolist()

print("Top classes conservées :", top_classes)

# Regrouper les autres en "other"
df["bug_type_cleaned"] = df["bug type"].apply(
    lambda x: x if x in top_classes else "other"
)

# Ré-encoder les classes en int, dans l’ordre de fréquence
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["bug_type_encoded"] = le.fit_transform(df["bug_type_cleaned"])

# Sauvegarder un nouveau fichier nettoyé
df.to_excel("classif_cleaned.xlsx", index=False)
print("✅ classif_cleaned.xlsx généré avec encodage bug_type_encoded")
```
